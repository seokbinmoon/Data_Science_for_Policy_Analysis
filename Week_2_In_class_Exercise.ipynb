# Part 2

This vector has the average high temperatures in Fort Collins for each month. Please run the line below.

```python
temperature_fc = tuple([44, 47, 55, 62, 71, 80, 87, 84, 76, 64, 51, 43])
# I used 'tuple' because each value is fixed(immutable).
```

Create a vector called "months" with elements Jan-Dec.

```python
months = tuple(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])
```

Assign the names of the months to the "temperature_fc" vector.

```python
temperature_fc_dict = dict(zip(months, temperature_fc))
temperature_fc_dict

# R: names(temperature_fc) <- months
```

Calculate and print the mean temperature in Fort Collins over the year.

```python
import numpy as np
np.mean(temperature_fc)

# R: print(mean(temperature_fc))
# In Python, we have to import 'numpy'
```

Create a logical vector called "above_freezing" that returns TRUE if the temperature of a month is above 32 degrees, FALSE otherwise.

```python
above_freezing = np.array(temperature_fc) > 32
above_freezing

# R: above_freezing <- temperature_fc > 32
# np.array converts the tuple to a NumPy array, enabling element-wise comparison (> 32).
```

This vector has the average high temperatures in Atalanta each month

```python
temperature_atlanta = tuple([52, 57, 65, 73, 80, 87, 89, 88, 82, 73, 63, 55])
```

Assign the names of the months to the "temperature_atlanta" vector.

```python
temperature_atlanta_dict = dict(zip(months, temperature_atlanta))
temperature_atlanta_dict
```

Generate a logical vector called "atlanta_warmer" that returns TRUE if Atlanta was warmer than Fort Collins on a given month.

```python
atlanta_warmer = np.array(temperature_fc) < np.array(temperature_atlanta)
atlanta_warmer

# R: atlanta_warmer <- temperature_fc < temperature_atlanta
```

Create a vector called "temperature_diff" that shows a difference in temperature between Atlanta and Fort Collins by month

```python
temperature_diff = np.array(temperature_atlanta) - np.array(temperature_fc)
temperature_diff
```

Define a variable called "temperature_march" that pulls the March value out of the temperature_diff variable.

```python
temperature_march = temperature_diff[2]
temperature_march
```
